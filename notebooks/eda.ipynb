# **Analise Exploratória de Dados**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/processed/dados_processed.csv')
df
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
sns.countplot(data=df, y=df['classe'])
plt.title("Distribuição de classes")
plt.xlabel("Classes")
plt.ylabel("Quantidade de classes")
plt.show();
```

```python
sns.scatterplot(data=df, x=df['altura'], y=df['largura'])
plt.title("Distribuição de altura e largura")
plt.xlabel("Altura")
plt.ylabel("largura")
plt.show();
```

```python
sns.histplot(data=df, x=df['Variancia_laplaciana'])
plt.title("Histograma da Variância laplaciana")
plt.xlabel("Valores da Variância")
plt.ylabel("Quantidade de Imagens (frequência)")
plt.show();
```

```python
sns.histplot(data=df, x=df['Aspect_Ratio'])
plt.title("Histograma do Aspect Ratio")
plt.xlabel("Valor da Proporção (Largura/Altura)")
plt.ylabel("Quantidade de Imagens (frequência)")
plt.show();
```

```python
sns.countplot(data=df, x=df['canais_cor'])
plt.title("Distribuição de cores")
plt.xlabel("Cores")
plt.ylabel("Quantidade de cores")
plt.show();
```

```python
sns.histplot(data=df, x=df['tamanho_kb'])
plt.title("Histograma de Tamanho dos arquivos")
plt.xlabel("Valor do tamanho")
plt.ylabel("Quantidade de Arquivo (frequência)")
plt.show();
```

```python
sns.histplot(data=df, x=df['brilho_medio'])
plt.title("Histograma do Brilho médio")
plt.xlabel("Valor do brilho")
plt.ylabel("Quantidade de Arquivo (frequência)")
plt.show();
```

```python
sns.histplot(data=df, x=df['contraste'])
plt.title("Histograma do Contraste")
plt.xlabel("Valor do contraste")
plt.ylabel("Quantidade de Arquivo (frequência)")
plt.show();
```

```python
df_duplicated = df[df['img_hash'].duplicated()]
df_duplicated
```

```python
agrupando_classe = df_duplicated.groupby('classe').size()
```

```python
agrupando_classe.plot(kind='barh')
plt.title("Duplicadas por classes")
plt.xlabel("Quantidade por classes")
plt.ylabel("Classes")
plt.show();
```
